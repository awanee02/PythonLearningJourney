Practice Program 1: CSV Reading & Analysis

Problem Statement: You are given a students.csv file. Write a program that:

Reads the CSV file

Prints all student records

Calculates average marks

Finds the top 3 students

Counts how many students passed (marks ≥ 40)

```python
import csv
import os

print("Student Record Analysis")

#Check if file exists
if not os.path.exists("students.csv"):
  print("The file 'students.csv' not found!")
  print("Please run the file creation code first.")
else:
  students = []
  with open("students.csv", "r") as file:
    reader = csv.DictReader(file)
    for row in reader:
      row['age'] = int(row['age'])
      row['marks'] = int(row['marks'])
      students.append(row)
      print(f"{row['name']} | Age: {row['age']} | Marks: {row['marks']}")

#Analysis
if students:
  total_marks = sum(student['marks'] for student in students)
  average = total_marks / len(students)
  passed = sum(1 for s in students if s['marks'] >= 40)
  print(f"Total students: {len(students)}")
  print(f"Average marks: {average}")
  print(f"Passed students: {passed}")

  #Top 3 students
  top_students = sorted(students, key=lambda x: x['marks'], reverse = True)[:3]
  print("\nTop 3 students: ")
  for i, student in enumerate(top_students, 1):
    print(f"{i}. {student['name']} - {student['marks']} marks")
```

Practice Program 2: JSON Read & Write


Problem Statement: Create a program that:

Creates a JSON file with product data

Reads it back

Adds a new product

Saves the updated data

```python
#JSON Read and Write
import json
import os

FILE_NAME = "products.json"

def load_products():
  if os.path.exists(FILE_NAME):
    try:
      with open(FILE_NAME, "r") as file:
        return json.load(file)
    except json.JSONDecodeError:
      print("JSON file is corrupted! Starting fresh.")
      return[]
  return[]

def save_products(products):
  try:
    with open(FILE_NAME, "w") as file:
      json.dump(products, file)
    print("Data saved successfully!")
  except Exception as e:
    print(f"Error saving file: {e}")

#Main program
print("JSON PRODUCT MANAGEMENT")

products = load_products()

#Add some default products if the file is empty
if not products:
  products = [
      {"id" : 1, "name" : "laptop", "price" : 65000, "category" : "electronics"},
      {"id" : 2, "name" : "Notebook", "price" : 150, "category" : "stationery"},
      {"id" : 3, "name" : "Smartphone", "price" : 45000, "category" : "electronics"}
  ]
  save_products(products)

#Display products
print("Current products: ")
for p in products:
  print(f"{p['name']} | {p['price']} | {p['category']}")

#Add a new product
print("Add New Product")
new_product = {
    "id" : len(products) + 1,
    "name" : input("Enter product name: "),
    "price" : float(input("Enter the product price: ")),
    "category" : input("Enter the product category: ")
}

products.append(new_product)
save_products(products)

print(f"\n New product '{new_product['name']} added successfully!")
```

Practice Program 3: CSV Data Processor

Problem Statement: Write a program that:

Takes sales data (product, quantity, price)

Saves it to CSV

Reads it and calculates total revenue

Finds the best-selling product

```python
#CSV Data Processor
import csv
import os
from datetime import datetime

FILE_NAME = "sales.csv"

def save_sales(sales_data):
  try:
    with open(FILE_NAME, "w") as file:
      writer = csv.writer(file)
      writer.writerow(["Date", "Product", "Quantity", "Unit_Price", "Total_amount"])
      writer.writerows(sales_data)
    print(f"Sales data saved to {FILE_NAME}")
  except Exception as e:
    print(f"Error saving file: {e}")
print("Sales Data Entry System \n")

sales = []
print("Enter sales records (type 'done' when finished.)\n")

while True:
  product = input("Product name (or 'done'): ")
  if product.lower() == 'done':
    break
  try:
    qty = int(input("Quantity: "))
    price = float(input("Price per unit: "))

    total = qty * price
    date = datetime.now().strftime("%Y-%m-%d")

    sales.append([date, product, qty, price, total])
    print(f"Added: {product} | Total : {total}\n")
  except ValueError:
    print("Invalid input! Please enter numbers correctly.")

if sales:
  save_sales(sales)

  #Analysis
  total_revenue = sum(row[4] for row in sales)
  best_seller = max(sales, key=lambda x: x[4])

  print("SALES SUMMARY")
  print(f"Total transactions: {len(sales)}")
  print(f"Total revenue: {total_revenue}")
  print(f"Best selling item: {best_seller[1]} ({best_seller[4]})")
else:
  print("No sales data entered.")
```

Mini Project for Day 20: Student Management System with CSV/JSON

Features:

Add new student

View all students

Search student by name

Save data to both CSV and JSON

Load data from file on startup

Show statistics (average, topper, pass/fail count)

```python
#Student Management System
import csv
import json
import os
from datetime import datetime

CSV_FILE = "students.csv"
JSON_FILE = "students.json"

class StudentManager:
  def __init__(self):
    self.students = []
    self.load_data()

  def load_data(self):
    if os.path.exists(JSON_FILE):
      try:
        with open(JSON_FILE, "r") as f:
          self.students = json.load(f)
        print(f"Loaded {len(self.students)} students from JSON.")
        return
      except:
        pass

    if os.path.exists(CSV_FILE):
      try:
        with open(CSV_FILE, "r") as f:
          reader = csv.DictReader(f)
          for row in reader:
            row['age'] = int(row['age'])
            row['marks'] = int(row['marks'])
            self.students.append(row)
        print(f"Loaded {len(self.students)} students from CSV.")
      except:
        self.students = []

  def save_data(self):
    try:
      #Save as JSON
      with open(JSON_FILE, "w") as f:
        json.dump(self.students, f)

      #Save as CSV
      with open(CSV_FILE, "w") as f:
        writer = csv.DictWriter(f, fieldnames=["name", "age", "marks"])
        writer.writeheader()
        writer.writerows(self.students)

        print("Data saved successfully to CSV and JSON!")
    except Exception as e:
      print(f"Error while saving: {e}")

  def add_student(self):
    try:
      name = input("Enter student name: ")
      age = int(input("Enter age: "))
      marks = int(input("Enter marks(0-100): "))

      if marks < 0 or marks > 100:
        print("Marks should be between 0 and 100")
        return

      student = {"name" : name, "age" : age, "marks" : marks}
      self.students.append(student)
      print(f"{name} added successfully!")
    except ValueError:
      print("Invalid input! Age and marks must be numbers.")

  def view_students(self):
    if not self.students:
      print("No students found!")
      return

    print(f"{'Name'} {'Age'} {'Marks'} Grade")
    for s in self.students:
      grade = "Pass" if s['marks'] >= 40 else "Fail"
      print(f"{s['name']} {s['age']} {s['marks']} {grade}")

  def search_students(self):
    keyword = input("Enter name to search: ")
    found = [s for s in self.students if keyword in s['name']]

    if found:
      print(f"Found {len(found)} students(s): ")
      for s in found:
        print(f"{s['name']} | Age: {s['age']} | Marks: {s['marks']}")
    else:
      print("No student found with that name.")

  def show_statistics(self):
    if not self.students:
      print("No data available!")
      return

    marks_list = [s['marks'] for s in self.students]
    print("STUDENT STATISTICS")
    print(f"Total Students : {len(self.students)}")
    print(f"Average Marks : {sum(marks_list)/len(marks_list):.2f}")
    print(f"Highest Marks : {max(marks_list)}")
    print(f"Lowest Marks : {min(marks_list)}")
    print(f"Pass Percentage : {sum(1 for m in marks_list if m >= 40) * 100 / len(marks_list):.1f}%")

def main():
  manager = StudentManager()
  print("Welcome to Student Management System")

  while True:
    print("1. Add new student")
    print("2. View all students")
    print("3. Search student")
    print("4. Show statistics")
    print("5. Save data")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':
      manager.add_student()
    elif choice == '2':
      manager.view_students()
    elif choice == '3':
      manager.search_student()
    elif choice == '4':
      manager.show_statistics()
    elif choice == '5':
      manager.save_data()
    elif choice == '6':
      manager.save_data()
      print("Thank you for using the system! Have a nice day.")
      break
    else:
      print("Invalid choice! Please try again.")

if __name__ == "__main__":
  main()
```
