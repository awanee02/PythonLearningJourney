Error Handling (Exception Handling)

```python
#Basic Syntax
try:
  number = int(input("Enter a number: "))
  result = 100 / number
  print(f"Result: {result}")
  #Code that might cause error
except ExceptionType as e:
  #Code that runs if error occurs:
  print("Error! Please enter a valid number.")
else:
  #Run only if no error occured
  print("Success!")
finally:
  #Always runs (cleanup code)
  print("This will always execute.")
```

```python
#Raising exceptions
try:
  age = float(input("Enter your age: "))
  if age < 0:
    raise ValueError("Age cannot be negative.")
  print(f"You are {age} years old.")

except ValueError as e:
  print(f"Invalid age: {e}")

except Exception as e:
  print(f"Unexpected error: {e}")

finally:
  print("Thank you for using this program.")
```

Exercise 1: Safe Calculator

Create a calculator that handles all possible errors gracefully:

Invalid input

Division by zero

Unsupported operations

```python
# Practice 1: Safe Calculator
while True:
    try:
        num1 = float(input("Enter first number: "))
        op = input("Enter operator (+, -, *, /): ")
        num2 = float(input("Enter second number: "))

        if op == "+":
            result = num1 + num2
        elif op == "-":
            result = num1 - num2
        elif op == "*":
            result = num1 * num2
        elif op == "/":
            if num2 == 0:
                raise ZeroDivisionError("Cannot divide by zero!")
            result = num1 / num2
        else:
            raise ValueError("Invalid operator! Use +, -, *, or / only.")

        print(f"Result: {num1} {op} {num2} = {result}")

    except ValueError as e:
        print(f"Invalid input: {e}")
    except ZeroDivisionError as e:
        print(f"Math Error: {e}")
    except Exception as e:
        print(f"Unexpected error: {e}")

    choice = input("\nDo you want to calculate again? (y/n): ")
    if choice != 'y':
        print("Thank you for using Safe Calculator!")
        break
```

Practice 2: Dictionary Safe Access

Write a program that:

Takes student name as input

Returns their marks safely (handle missing student)

Allows adding new students

Handles invalid marks input

```python
students = {
    'Awanee' : 98,
    'Arnav' : 96,
    'Anay' : 87,
    'Vedika' : 82,
    'Parth' : 85
}

print("--------------Student Marks Calculator-----------------")

while True:
  try:
    print("Options: ")
    print("1. Check marks")
    print("2. Add/update student")
    print("3. Show all students")
    print("4. Exit")

    choice = int(input("Enter your choice (1-4): "))

    if choice == 1:
      name = input("Enter a student name: ")
      marks = students.get(name)
      if marks is not None:
        print(f"{name.title()} has {marks} marks.")
      else:
        print(f"{name.title()} name not found!")

    elif choice == 2:
      name = input("Enter a student name: ")
      marks = int(input("Enter the marks (0-100): "))
      if marks < 0 or marks > 100:
        raise ValueError("Marks should be between 0 and 100.")
      students[name] = marks
      print(f"{name.title()} with {marks} marks updated.")

    elif choice == 3:
      if not students:
        print("No students found!")
      else:
        print("All students: ")
        for name, marks in students.items():
          print(f"{name.title()} : {marks}")

    elif choice == 4:
      print("Goodbye!")
      break

    else:
      print("Invalid choice, please select (1-4): ")

  except ValueError as e:
    print(f"Invalid input: {e}")
  except Exception as e:
    print(f"Unexpected error: {e}")
```

Practice 3: Robust Number List Processor

Take a list of strings from user, convert them to integers, and:

Calculate sum, average, max, min

Handle non-numeric values gracefully

Handle empty list case

```python
while True:
  try:
    user_input = input("Enter numbers separated by comma: ")
    items = [item.strip() for item in user_input.split(",")]

    numbers = []
    errors = []

    for item in items:
      try:
        num = float[item]
        numbers.append(num)
      except ValueError:
        errors.append(item)

    if not numbers:
      raise ValueError("No valid numbers entered.")

    print(f"Valid numbers: {numbers}")
    if errors:
      print(f"Ignored invalid values: {errors}")

    print("Results: ")
    print(f"Sum = {sum(numbers)}")
    print(f"Average = {sum(numbers) / len(numbers)}")
    print(f"Minimum = {min(numbers)}")
    print(f"Maximum = {max(numbers)}")
    print(f"Count = {len(numbers)}")

  except ValueError as e:
    print(f"Error: {e}")
  except Exception as e:
    print(f"Unexpected error: {e}")

  again = input("Process another list? (y/n): ")
  if again != 'y':
    print("Thank you!")
    break
```
