Bank Account Simulator (using OOP)

This is a pure project day to strengthen your OOP skills. You will build a realistic Bank Account System using classes, inheritance, and everything you’ve learned so far.

Create a Banking System with two types of accounts:

Savings Account

Current Account

Both will inherit from a base BankAccount class.

```python
#Bank Account Simulator

from datetime import datetime

class BankAccount:
    def __init__(self, account_number, account_holder, balance=0.0):
        self.account_number = account_number
        self.account_holder = account_holder
        self.balance = balance
        self.transactions = []  # List to store transaction history

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            self.transactions.append({
                "date": datetime.now().strftime("%Y-%m-%d %H:%M"),
                "type": "Deposit",
                "amount": amount,
                "balance": self.balance
            })
            print(f"{amount:,} deposited successfully!")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance!")
        elif amount <= 0:
            print("Amount must be positive.")
        else:
            self.balance -= amount
            self.transactions.append({
                "date": datetime.now().strftime("%Y-%m-%d %H:%M"),
                "type": "Withdrawal",
                "amount": amount,
                "balance": self.balance
            })
            print(f"{amount:,} withdrawn successfully!")

    def check_balance(self):
        print(f"\nAccount Balance: ₹{self.balance:,.2f}")

    def show_transaction_history(self):
        if not self.transactions:
            print("No transactions yet.")
            return
        print(f"Transaction History for {self.account_holder}")
        for t in self.transactions:
            print(f"{t['date']} | {t['type']:10} | ₹{t['amount']:10,.2f} | Balance: ₹{t['balance']:,.2f}")

    def __str__(self):
        return f"{self.account_number} | {self.account_holder} | Balance: ₹{self.balance:,.2f}"


class SavingsAccount(BankAccount):
    def __init__(self, account_number, account_holder, balance=0.0, interest_rate=3.5):
        super().__init__(account_number, account_holder, balance)
        self.interest_rate = interest_rate  # in percentage

    def add_interest(self):
        interest = (self.balance * self.interest_rate) / 100
        self.balance += interest
        self.transactions.append({
            "date": datetime.now().strftime("%Y-%m-%d %H:%M"),
            "type": "Interest",
            "amount": interest,
            "balance": self.balance
        })
        print(f"Interest of ₹{interest:,.2f} added at {self.interest_rate}% rate!")

    def __str__(self):
        return super().__str__() + f" | Type: Savings | Rate: {self.interest_rate}%"


class CurrentAccount(BankAccount):
    def __init__(self, account_number, account_holder, balance=0.0, overdraft_limit=5000):
        super().__init__(account_number, account_holder, balance)
        self.overdraft_limit = overdraft_limit

    def withdraw(self, amount):
        max_withdraw = self.balance + self.overdraft_limit
        if amount > max_withdraw:
            print(f"Overdraft limit exceeded! Max you can withdraw: ₹{max_withdraw:,}")
        elif amount <= 0:
            print("Amount must be positive.")
        else:
            self.balance -= amount
            self.transactions.append({
                "date": datetime.now().strftime("%Y-%m-%d %H:%M"),
                "type": "Withdrawal",
                "amount": amount,
                "balance": self.balance
            })
            print(f"{amount:,} withdrawn successfully!")
            if self.balance < 0:
                print(f"You are in overdraft by ₹{abs(self.balance):,.2f}")

    def __str__(self):
        return super().__str__() + f" | Type: Current | Overdraft: ₹{self.overdraft_limit:,}"


# ==================== MAIN PROGRAM ====================
def main():
    accounts = {}
    print("Welcome to Python Bank Simulator\n")

    while True:
        print("1. Create Savings Account")
        print("2. Create Current Account")
        print("3. Deposit Money")
        print("4. Withdraw Money")
        print("5. Check Balance")
        print("6. Show Transaction History")
        print("7. Add Interest (Savings only)")
        print("8. Show All Accounts")
        print("9. Exit")

        choice = input("\nEnter your choice (1-9): ").strip()

        try:
            if choice == "1":  # Savings Account
                acc_no = input("Enter Account Number: ")
                name = input("Enter Account Holder Name: ")
                bal = float(input("Enter Initial Balance: "))
                rate = float(input("Enter Interest Rate (%): "))
                accounts[acc_no] = SavingsAccount(acc_no, name, bal, rate)
                print("Savings Account created successfully!")

            elif choice == "2":  # Current Account
                acc_no = input("Enter Account Number: ")
                name = input("Enter Account Holder Name: ")
                bal = float(input("Enter Initial Balance: "))
                limit = float(input("Enter Overdraft Limit: "))
                accounts[acc_no] = CurrentAccount(acc_no, name, bal, limit)
                print("Current Account created successfully!")

            elif choice in ["3", "4", "5", "6", "7"]:
                acc_no = input("Enter Account Number: ")
                if acc_no not in accounts:
                    print("Account not found!")
                    continue
                acc = accounts[acc_no]

                if choice == "3":
                    amt = float(input("Enter deposit amount: "))
                    acc.deposit(amt)
                elif choice == "4":
                    amt = float(input("Enter withdrawal amount: "))
                    acc.withdraw(amt)
                elif choice == "5":
                    acc.check_balance()
                elif choice == "6":
                    acc.show_transaction_history()
                elif choice == "7":
                    if isinstance(acc, SavingsAccount):
                        acc.add_interest()
                    else:
                        print("Interest is only available for Savings Account!")

            elif choice == "8":
                if not accounts:
                    print("No accounts created yet.")
                else:
                    print("\nAll Accounts:")
                    for acc in accounts.values():
                        print(acc)

            elif choice == "9":
                print("Thank you for using Python Bank Simulator!")
                break

            else:
                print("Invalid choice! Please select 1-9.")

        except ValueError:
            print("Please enter valid numbers.")
        except Exception as e:
            print(f"Error: {e}")


if __name__ == "__main__":
    main()
```
