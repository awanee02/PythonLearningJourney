Practice Program 1: Random Module

Problem Statement: Create a program that:

Generates 5 random numbers between 1 and 100

Picks a random name from a list of 6 names

Shuffles a list of numbers [1,2,3,4,5,6] and prints it

Generates a random OTP of 6 digits

```python
#Random Module
import random

print('-------------RANDOM MODULE MENU---------------')

#1. Random Numbers
print("Any 5 random numbers (1-100): ")
for _ in range(5):
  print(random.randint(1, 100), end = ' ')
print("\n")

#2. Random choice from list
names = ["Awanee", "Arnav", "Parth", "Vedika", "Anay", "Spruha"]
print("Randomly selected name: ", random.choice(names))

#3. Shuffle the list of numbers
numbers = [1,2,3,4,5,6]
random.shuffle(numbers)
print("Shuffled list: ", numbers)

#4. Generate a random OTP of 6 digits
otp = ''.join(str(random.randint(0,9)) for _ in range(6))
print("Your OTP: ", otp)
```

Practice Program 2: Datetime Module

Problem Statement: Write a program that:

Prints current date and time in different formats

Calculates your age in years, months, and days (take your birthdate as input)

Shows how many days are left until New Year (1st Jan 2027)

```python
#Datetime Module
from datetime import datetime, date

print("---------------DATETIME MODULE DEMO------------------")

#Current date and time
now = datetime.now()
print("Current date and time: ", now)
print("Formatted: ", now.strftime("%d-%m-%Y %H:%M:%S"))
print("Only date: ", now.date())
print("Only time: ", now.time())

#Age Calculator
birth_str = input("Enter your date of birth (YYYY-MM-DD): ")
birth_date = datetime.strptime(birth_str, "%Y-%m-%d").date()
today = date.today()

age = today.year - birth_date.year - ((today.month, today.day) < (birth_date.month, birth_date.day))
print(f"Your age is {age} years.")

#Days until new year 2027
new_year = date(2027, 1, 1)
days_left = (new_year - today).days
print(f"Days left until new year 2027 are: {days_left} days.")
```

Practice Program 3: OS Module

Problem Statement: Create a program that:

Shows current working directory

Lists all files and folders in current directory

Creates a new folder called My_Project

Checks if a file named report.txt exists

Prints file size of my_diary.txt (from yesterday) if it exists

```python
#OS Module
import os

print("----------OS MODULE DEMO------------\n")

#Current Directory
print("Current working directory: ", os.getcwd())

#List files and folders in current directory
print("Files and folders in current directory: ")
print(os.listdir())

#Create folder
folder_name = "My Project"
if not os.path.exists(folder_name):
  os.mkdir(folder_name)
  print(f"Folder '{folder_name}' created!")
else:
  print(f"Folder '{folder_name}' already exists.\n")

#Check if file exists
if os.path.exists("my_journal.txt"):
  size = os.path.getsize("my_journal.txt") / 1024
  print(f"my_journal.txt exists | Size: {size: .2f} KB.")
else:
  print("my_journal.txt does not exist.")
```

Mini Project for Day 16: Smart To-Do List Manager

Project Requirements:

Build a feature-rich To-Do List that uses all three modules (random, datetime, os).

Features:

Add a new task (with due date)

View all tasks

Mark task as completed

Delete a task

Show overdue tasks

"Motivational Quote of the Day" (using random)

Save and load tasks from a file automatically

```python
# Mini Project: Smart To-Do List Manager
import random
import datetime
import os
import json

TODO_FILE = "todo_list.json"

quotes = [
    "The best way to get started is to quit talking and begin doing.",
    "Small daily improvements are the key to staggering long-term results.",
    "You don't have to be great to start, but you have to start to be great.",
    "Focus on progress, not perfection."
]

def load_tasks():
    if os.path.exists(TODO_FILE):
        try:
            with open(TODO_FILE, "r") as f:
                return json.load(f)
        except:
            return []
    return []

def save_tasks(tasks):
    with open(TODO_FILE, "w") as f:
        json.dump(tasks, f, indent=4)

def add_task(tasks):
    task = input("Enter task: ")
    due_date = input("Enter due date (YYYY-MM-DD) or press Enter: ")
    tasks.append({
        "task": task,
        "due": due_date if due_date else None,
        "status": "Pending"
    })
    print("Task added!")

def view_tasks(tasks):
    if not tasks:
        print("No tasks yet!")
        return
    print("\nYour Tasks:")
    for i, t in enumerate(tasks, 1):
        due = t['due'] if t['due'] else "No due date"
        print(f"{i}. [{t['status']}] {t['task']} | Due: {due}")

def main():
    tasks = load_tasks()
    print("Welcome to Smart To-Do List!")

    while True:
        print("\n" + "="*50)
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Mark as Completed")
        print("4. Delete Task")
        print("5. Motivational Quote")
        print("6. Exit")
        print("="*50)

        try:
            choice = int(input("\nEnter choice: "))

            if choice == 1:
                add_task(tasks)
            elif choice == 2:
                view_tasks(tasks)
            elif choice == 3:
                view_tasks(tasks)
                idx = int(input("Enter task number to complete: ")) - 1
                if 0 <= idx < len(tasks):
                    tasks[idx]["status"] = "Completed"
                    print("Task marked completed!")
            elif choice == 4:
                view_tasks(tasks)
                idx = int(input("Enter task number to delete: ")) - 1
                if 0 <= idx < len(tasks):
                    del tasks[idx]
                    print("Task deleted!")
            elif choice == 5:
                print("\n Quote of the Day:")
                print(random.choice(quotes))
            elif choice == 6:
                save_tasks(tasks)
                print("Goodbye! Stay productive!C")
                break
            else:
                print("Invalid choice!")

        except Exception as e:
            print("Invalid input. Please try again.")

if __name__ == "__main__":
    main()
```
