Practice Program 1: Single Inheritance

Problem Statement: Create a Person parent class with name and age.

Create a Student child class that inherits from Person and adds marks and course.

Add a method get_details() in both classes and use super().

```python
#Single Inheritance

class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def get_details(self):
    print(f"Name: {self.name}, Age: {self.age}")

class Student(Person):
  def __init__(self, name, age, marks, course):
    super().__init__(name, age)    #Calling parent constructor
    self.marks = marks
    self.course = course

  def get_details(self):
    super().get_details()   #Calling parent method
    print(f"Course: {self.course}, Marks: {self.marks}")
    print(f"Grade: {'Pass' if self.marks >= 40 else 'Fail'}")

#Testing
s1 = Student('Awanee Patunkar', 22, 95, 'Python Learning')
s1.get_details()
```

Practice Program 2: Magic Methods (__str__, __len__, __add__)

Problem Statement: Create a Book class with title, author, and pages.
Implement:

__str__() → nice readable output

__len__() → returns number of pages

__add__() → adds pages of two books

```python
#Magic Methods

class Book:
  def __init__(self, title, author, pages):
    self.title = title
    self.author = author
    self.pages = pages

  def __str__(self):
    return f"'{self.title}' by {self.author} ({self.pages} pages.)"

  def __len__(self):
    return self.pages

  def __add__(self, other):
    return f"Combined pages: {self.pages + other.pages}"

book1 = Book("Python Mastery", "Guido van Rossum", 450)
book2 = Book("Clean Code", "Robert Martin", 930)
print(book1)
print(f"Number of pages in book1: {len(book1)}")
print(book1 + book2)
```

Practice Program 3: Multilevel Inheritance

Problem Statement: Create three classes:

Vehicle (parent)

Car (inherits from Vehicle)

ElectricCar (inherits from Car)

Add appropriate attributes and methods in each level.

```python
#Multilevel Inheritance

class Vehicle:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def start(self):
    print(f"{self.brand} {self.model} is starting...")

class Car(Vehicle):
  def __init__(self, brand, model, fuel_type):
    super().__init__(brand, model)
    self.fuel_type = fuel_type

  def drive(self):
    print(f"Driving {self.model} on {self.fuel_type}.")

class ElectricCar(Car):
  def __init__(self, brand, model, battery_capacity):
    super().__init__(brand, model, "Electricity")
    self.battery_capacity = battery_capacity

  def drive(self):
    print(f"Driving {self.model} on {self.battery_capacity} kWh battery and it moves silently.")

#Testing
tesla = ElectricCar("Tesla", "Model S", 100)
tesla.start()
tesla.drive()
```

Mini Project for Day 18: Employee Management System

Project Requirements: Create a system with:

Employee (base class)

Manager (inherits from Employee)

Developer (inherits from Employee)

Features to implement:

Add different types of employees

Display employee details (use __str__)

Calculate salary (different logic for Manager & Developer)

Give bonus (polymorphism)

```python
#Employee Management System
class Employee:
    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id
        self.name = name
        self.salary = salary

    def __str__(self):
        return f"ID: {self.emp_id} | Name: {self.name} | Salary: ₹{self.salary:,}"

    def give_bonus(self, amount):
        if amount > 0:
            self.salary += amount
            print(f"Bonus of ₹{amount:,} given to {self.name}")
        else:
            print("Bonus amount must be positive!")


class Manager(Employee):
    def __init__(self, emp_id, name, salary, department):
        super().__init__(emp_id, name, salary)
        self.department = department
        self.team_size = 0

    def __str__(self):
        return super().__str__() + f" | Department: {self.department} | Team Size: {self.team_size}"

    def give_bonus(self, amount):
        bonus = amount + 5000  # Managers get extra ₹5000 bonus
        print(f"Manager bonus applied (extra ₹5000)")
        super().give_bonus(bonus)

    def add_team_member(self):
        self.team_size += 1
        print(f"Team member added. Current team size: {self.team_size}")


class Developer(Employee):
    def __init__(self, emp_id, name, salary, programming_lang):
        super().__init__(emp_id, name, salary)
        self.programming_lang = programming_lang
        self.projects_completed = 0

    def __str__(self):
        return super().__str__() + f" | Language: {self.programming_lang} | Projects: {self.projects_completed}"

    def complete_project(self):
        self.projects_completed += 1
        print(f"Project completed! Total projects: {self.projects_completed}")


#MAIN PROGRAM
def main():
    employees = []
    print("Welcome to Employee Management System\n")

    while True:
        print("1. Add Manager")
        print("2. Add Developer")
        print("3. Show All Employees")
        print("4. Give Bonus")
        print("5. Complete Project (Developer only)")
        print("6. Add Team Member (Manager only)")
        print("7. Exit")

        try:
            choice = input("\nEnter your choice (1-7): ")

            if choice == "1":
                emp_id = input("Enter Employee ID: ")
                name = input("Enter Name: ")
                salary = int(input("Enter Salary: "))
                department = input("Enter Department: ")
                manager = Manager(emp_id, name, salary, department)
                employees.append(manager)
                print(f"Manager '{name}' added successfully!")

            elif choice == "2":
                emp_id = input("Enter Employee ID: ")
                name = input("Enter Name: ")
                salary = int(input("Enter Salary: "))
                lang = input("Enter Programming Language: ")
                dev = Developer(emp_id, name, salary, lang)
                employees.append(dev)
                print(f"Developer '{name}' added successfully!")

            elif choice == "3":
                if not employees:
                    print("No employees added yet.")
                else:
                    print("ALL EMPLOYEES")
                    for i, emp in enumerate(employees, 1):
                        print(f"{i}. {emp}")

            elif choice == "4":
                if not employees:
                    print("No employees to give bonus.")
                    continue
                for i, emp in enumerate(employees, 1):
                    print(f"{i}. {emp.name} ({type(emp).__name__})")
                idx = int(input("\nEnter employee number: ")) - 1
                amount = int(input("Enter bonus amount: "))
                if 0 <= idx < len(employees):
                    employees[idx].give_bonus(amount)
                else:
                    print("Invalid employee number!")

            elif choice == "5":
                devs = [emp for emp in employees if isinstance(emp, Developer)]
                if not devs:
                    print("No developers found!")
                    continue
                for i, dev in enumerate(devs, 1):
                    print(f"{i}. {dev.name}")
                idx = int(input("\nEnter developer number: ")) - 1
                if 0 <= idx < len(devs):
                    devs[idx].complete_project()

            elif choice == "6":
                managers = [emp for emp in employees if isinstance(emp, Manager)]
                if not managers:
                    print("No managers found!")
                    continue
                for i, mgr in enumerate(managers, 1):
                    print(f"{i}. {mgr.name}")
                idx = int(input("\nEnter manager number: ")) - 1
                if 0 <= idx < len(managers):
                    managers[idx].add_team_member()

            elif choice == "7":
                print("Thank you for using Employee Management System! 👋")
                break

            else:
                print("Invalid choice! Please select 1-7.")

        except ValueError:
            print("Please enter valid numbers where required.")
        except Exception as e:
            print(f"Error: {e}")


if __name__ == "__main__":
    main()
```
