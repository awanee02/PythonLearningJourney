Practice Program 1: Basic Class

Problem Statement: Create a Student class with the following:

Attributes: name, age, marks

Method get_details() → prints student information

Method is_passed() → returns True if marks >= 40

Create 3 student objects and show their details

```python
class Student:
  def __init__(self, name, age, marks):
    self.name = name
    self.age = age
    self.marks = marks

  def get_details(self):
    print(f"Name: {self.name}, Age: {self.age}, Marks: {self.marks}")

  def is_passed(self):
    return self.marks >= 40

#Creating objects
s1 = Student("Awanee", 22, 98)
s2 = Student("Arnav", 16, 99)
s3 = Student("Parth", 18, 94)

s1.get_details()
s2.get_details()
s3.get_details()

print(f"Is {s1.name} passed? {s1.is_passed()}")
print(f"Is {s3.name} passed? {s3.is_passed()}")
```

Practice Program 2: Bank Account Class

Problem Statement: Create a BankAccount class that has:

Attributes: account_holder, balance (default 0)

Methods:
deposit(amount)

withdraw(amount) (should not allow if insufficient balance)

check_balance()

```python
#Bank Account Class
class BankAccount:
  def __init__(self, account_holder, balance = 0):
    self.account_holder = account_holder
    self.balance = balance

  def deposit(self, amount):
    if amount > 0:
      self.balance += amount
      print(f"Deposited Rs.{amount}. Your new balance is {self.balance}.")
    else:
      print("Amount must be positive.")

  def withdraw(self, amount):
    if amount > self.balance:
      print("Insufficient balance!")
    elif amount <= 0:
      print("Amount must be positive.")
    else:
      self.balance -= amount
      print(f"Wthdrawn amount is Rs. {amount}. Your remaining balance is {self.balance}.")

  def check_balance(self):
    print(f"Current balance for {self.account_holder} is {self.balance}.")

#Creating object
acc = BankAccount("Awanee Patunkar", 10000000)
acc.check_balance()
acc.deposit(200000)
acc.withdraw(50000)
```

Practice Program 3: Rectangle & Circle Class

Problem Statement: Create two classes:

Rectangle → attributes length, width. Methods: area(), perimeter()

Circle → attribute radius. Methods: area(), circumference()

Use proper naming and print results for sample values.

```python
import math

#Rectangle and Circle classes

class Rectangle:
  def __init__(self, length, width):
    self.length = length
    self.width = width

  def area(self):
    return self.length * self.width

  def perimeter(self):
    return 2 * (self.length + self.width)

class Circle:
  def __init__(self, radius):
    self.radius = radius

  def area(self):
    return math.pi * (self.radius ** 2)

  def circumference(self):
    return 2 * math.pi * self.radius

#Creating objects
r1 = Rectangle(4,5)
print(f"Area of rectangle: {r1.area()}")
print(f"Perimeter of rectangle: {r1.perimeter()}")

c1 = Circle(3)
print(f"Area of circle: {c1.area()}")
print(f"Circumference: {c1.circumference()}")
```

Mini Project for Day 17: Library Management System

Project Name: Personal Library Manager

Requirements: Create a Book class and a Library class.

Book Class:

Attributes: title, author, isbn, status ("Available" / "Borrowed")

Library Class:

Method to add_book()

Method to borrow_book(title)

Method to return_book(title)

Method to show_available_books()

Make a menu-driven program.

```python
#Library Management System
class Book:
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.status = "Available"

    def __str__(self):
        return f"{self.title} by {self.author} (ISBN: {self.isbn}) - {self.status}"


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, title, author, isbn):
        book = Book(title, author, isbn)
        self.books.append(book)
        print(f"Book '{title}' added successfully!")

    def show_available_books(self):
        print("\n Available Books:")
        available = [book for book in self.books if book.status == "Available"]
        if not available:
            print("No books available.")
        else:
            for i, book in enumerate(available, 1):
                print(f"{i}. {book}")

    def borrow_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower() and book.status == "Available":
                book.status = "Borrowed"
                print(f"You borrowed '{book.title}'")
                return
        print("Book not available or doesn't exist.")

    def return_book(self, title):
        for book in self.books:
            if book.title.lower() == title.lower() and book.status == "Borrowed":
                book.status = "Available"
                print(f"Thank you for returning '{book.title}'")
                return
        print("This book was not borrowed or doesn't exist.")


# Main Program
def main():
    lib = Library()
    print("Welcome to Personal Library System\n")

    while True:
        print("\n" + "="*40)
        print("1. Add Book")
        print("2. Show Available Books")
        print("3. Borrow Book")
        print("4. Return Book")
        print("5. Exit")
        print("="*40)

        try:
            choice = int(input("\nEnter your choice: "))

            if choice == 1:
                title = input("Enter Book Title: ")
                author = input("Enter Author Name: ")
                isbn = input("Enter ISBN: ")
                lib.add_book(title, author, isbn)

            elif choice == 2:
                lib.show_available_books()

            elif choice == 3:
                title = input("Enter title of book to borrow: ")
                lib.borrow_book(title)

            elif choice == 4:
                title = input("Enter title of book to return: ")
                lib.return_book(title)

            elif choice == 5:
                print("Thank you for using Library System!")
                break

            else:
                print("Invalid choice!")

        except ValueError:
            print("
            Please enter a valid number.")

if __name__ == "__main__":
    main()
```
