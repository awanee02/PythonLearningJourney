```python
#Basic File Operations

def basic_file_operations():
  #Writing to a file
  with open ("info.txt", "w") as f:
    f.write("Heyaaa, I am learning Python!")
    f.write("Today is Day 15 - File Handling.")
    f.writelines(["Python is actually cool and file handling is important."])

  #Reading the file
  print("------------FILE CONTENT------------------")
  with open("info.txt", "r") as f:
    content = f.read()
    print(content)

  #Reading line by line
  print("-----------LINE BY LINE------------")
  with open("info.txt", "r") as f:
    for line in f:
      print(line.strip())

basic_file_operations()
```

```python
#Student Data Manager
def student_data_manager():
  students = [
      {"name" : "awanee", "age" : 22, "marks" : 98},
      {"name" : "arnav", "age" : 16, "marks" : 96},
      {"name" : "vedika", "age" : 13, "marks" : 86}
  ]

  #Write to file
  with open("students.txt", "w") as f:
    f.write("Name, age, marks \n")
    for student in students:
      f.write(f"{student['name']}, {student['age']}, {student['marks']} \n")

  #Read and display
  print("------------STUDENT RECORDS---------------")
  with open("students.txt", "r") as f:
    header = f.readline()
    print(header.strip())
    for line in f:
      name, age, marks = line.strip().split(",")
      print(f"Name : {name} | Age : {age} | Marks : {marks}")

student_data_manager()
```

```python
# Practice 3: Word Counter with Logging
import datetime

def word_counter():
    text = """Python is a powerful programming language.
File handling helps us work with data persistently.
Error handling and file handling are very important concepts."""

    # Write sample text
    with open("sample.txt", "w") as f:
        f.write(text)

    # Read and count words
    with open("sample.txt", "r") as f:
        content = f.read()
        words = content.split()
        word_count = len(words)
        line_count = len(content.splitlines())

    print(f"Total Words : {word_count}")
    print(f"Total Lines : {line_count}")

    # Logging
    with open("activity.log", "a") as log:
        log.write(f"[{datetime.datetime.now()}] Analyzed file - Words: {word_count}\n")

word_counter()
```

Project Name: My Personal Diary

Features to Implement:

Add New Entry (with date & time automatically)

View All Entries

View Entries by Date

Delete All Entries (with confirmation)

Proper error handling and user-friendly menu

```python
#Personal Journal
import datetime
import os

JOURNAL_FILE = "my_journal_txt"

def add_entry():
  print("--------------NEW JOURNAL ENTRY---------------")
  entry = input("Enter your thoughts: \n")
  timestamp = datetime.datetime.now().strftime

  with open("JOURNAL_FILE" "a") as f:
    f.write(f"Date and time : {timestamp} \n")
    f.write(f"{entry} \n")
  print("Entry saved successfully.")

def view_entries():
  if not os.path.exists(JOURNAL_FILE):
    print("No entries yet. Start writing your first one!")
    return

  with open("JOURNAL_FILE", "r") as f:
    content = f.read()
    if content.strip():
      print("              MY DIARY            ")
      print(content)
    else:
      print("Your diary is empty.")

def clear_journal():
  if os.path.exists(JOURNAL_FILE):
    confirm = input("Are you sure you want to clear ALL entries? (y/n): ").lower()
    if confirm == 'yes':
      os.remove(JOURNAL_FILE)
      print("All entries deleted.")
    else:
      print("Operation cancelled!")
  else:
    print("Journal is already empty!")

#Main menu
while True:
  try:
    print("----------MENU-------------")
    print("1. Add a new entry")
    print("2. View all entries")
    print("3. Clear journal")
    print("4. Exit")

    choice = int(input("Enter your choice (1-4) : "))

    if choice == 1:
      add_entry()
    elif choice == 2:
      view_entries()
    elif choice == 3:
      clear_journal()
    elif choice == 4:
      print("Goodbye!")
      break
    else:
      print("Invalid choice! Please select 1-4")

  except ValueError:
    print("Please enter a valid number.")
  except Exception as e:
    print(f"Error : {e}")
```
